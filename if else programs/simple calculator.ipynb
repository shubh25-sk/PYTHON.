#Simple clculator

```python
a = float(input("enter the first number:"))
b = float(input("enter the second number:"))

print("Select operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

z = input("Enter choice (1/2/3/4): ")

if z == '1':
    result = a + b
    print("Result:",result)
elif z == '2':
    result = a - b
    print("Result:",result)
elif z == '3':
    result = a * b
    print("Result:",result)
elif z == '4':
    if b != 0:
        result = a / b
        print("Result:",result)
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Invalid choice. Please enter 1, 2, 3, or 4.")
```
